```python
from collections import deque

# Graph represented as an adjacency list
graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": ["G"],
    "F": [],
    "G": []
}

# BFS Function
def bfs(graph, start):
    visited = set()
    queue = deque([start])
    traversal = []

    while queue:
        node = queue.popleft()

        if node not in visited:
            visited.add(node)
            traversal.append(node)

            # Add all unvisited neighbors to the queue
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(neighbor)

    return traversal

# Driver Code
start_node = "A"
result = bfs(graph, start_node)

print("BFS Traversal:")
print(" -> ".join(result))
```

```python
#Dfs of this above code

import networkx as nx
import matplotlib.pyplot as plt

# Graph Definition
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': [],
    'E': [],
    'F': ['G'],
    'G': []
}

# DFS Function
def dfs(graph, node, visited=None, traversal=None):
    if visited is None:
        visited = set()
    if traversal is None:
        traversal = []

    visited.add(node)
    traversal.append(node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited, traversal)

    return traversal

# Perform DFS
result = dfs(graph, 'A')
print("DFS Traversal:", " -> ".join(result))

# Create NetworkX Directed Graph
G = nx.DiGraph(graph)

plt.figure(figsize=(8, 6))

# Node Positions
pos = {
    'A': (0, 2),
    'B': (-2, 1),
    'C': (2, 1),
    'D': (-3, 0),
    'E': (-1, 0),
    'F': (1, 0),
    'G': (3, 0)
}

# Draw Graph
nx.draw_networkx_nodes(
    G,
    pos,
    node_size=1200,
    node_color="lightgreen",
    edgecolors="yellow"
)

nx.draw_networkx_labels(
    G,
    pos,
    font_size=12,
    font_weight="bold"
)

nx.draw_networkx_edges(
    G,
    pos,
    arrows=True,
    arrowstyle="-|>",
    arrowsize=20,
    edge_color="black",
    width=2
)

plt.title("Graphical Representation of DFS", fontsize=14)
plt.axis("off")
plt.show()
```
