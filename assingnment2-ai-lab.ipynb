mat-clot for visualisation 

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

digits = load_digits()

n = len(digits.images)

cols = 10
rows = (n + cols - 1) // cols

plt.figure(figsize=(15, rows * 1.5))

for i in range(10):
    plt.subplot(rows, cols, i + 1)
    plt.imshow(digits.images[i], cmap='gray')
    plt.title(digits.target[i], fontsize=6)
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

# Load the dataset
digits = load_digits()

# Display the first 10 images
plt.figure(figsize=(12, 3))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(digits.images[i], cmap='gray')
    plt.title(f"Label: {digits.target[i]}")
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

# Load the dataset
digits = load_digits()

# Create a 5x2 grid (10 images)
fig, axes = plt.subplots(5, 2, figsize=(10, 10))

# Display the first 10 images
for i, ax in enumerate(axes.flat):
    ax.imshow(digits.images[i], cmap='gray')
    ax.set_title(f"Digit: {digits.target[i]}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```
