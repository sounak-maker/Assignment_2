```python
# Develop  a python program to implement a Rule-Based Expert System for Student Performance Evaluation.The system should take inputs such as attendance,marks,assignment performance,and internal assessment and use IF THEN rules to classify the student as Excellent,Good,Average,or Poor.

# Rule-Based Expert System for Student Performance Evaluation

def evaluate_student(attendance, marks, assignment, internal):
    
    # Rule 1: Excellent
    if attendance >= 90 and marks >= 85 and assignment >= 85 and internal >= 85:
        performance = "Excellent"

    # Rule 2: Good
    elif attendance >= 75 and marks >= 70 and assignment >= 70 and internal >= 70:
        performance = "Good"

    # Rule 3: Average
    elif attendance >= 60 and marks >= 50 and assignment >= 50 and internal >= 50:
        performance = "Average"

    # Rule 4: Poor
    else:
        performance = "Poor"

    return performance


# Taking input from the user
print("===== Student Performance Evaluation System =====")

attendance = float(input("Enter attendance percentage: "))
marks = float(input("Enter marks percentage: "))
assignment = float(input("Enter assignment performance percentage: "))
internal = float(input("Enter internal assessment percentage: "))

# Evaluate performance
result = evaluate_student(attendance, marks, assignment, internal)

# Display result
print("\n===== Evaluation Result =====")
print("Attendance:", attendance, "%")
print("Marks:", marks, "%")
print("Assignment:", assignment, "%")
print("Internal Assessment:", internal, "%")
print("Overall Performance:", result)
```

```python
# Describe a python to implement a rule-based expert system for medical diagnosis.The system should accept symptoms such as fever,cough,hedache,and sore throat and use IF THEN rule to identify a possible disease.

# Rule-Based Expert System for Medical Diagnosis

def diagnose(fever, cough, headache, sore_throat):

    # Rule 1: Flu
    if fever and cough and headache and sore_throat:
        disease = "Flu"

    # Rule 2: Common Cold
    elif cough and sore_throat and not fever:
        disease = "Common Cold"

    # Rule 3: Fever with headache
    elif fever and headache and not cough:
        disease = "Possible Viral Fever"

    # Rule 4: Sore throat and cough
    elif sore_throat and cough:
        disease = "Possible Throat Infection"

    # Rule 5: No matching rule
    else:
        disease = "Disease cannot be identified from the given symptoms"

    return disease


# Taking input from the user
print("===== Medical Diagnosis Expert System =====")
print("Answer with Yes or No\n")

fever = input("Do you have fever? ").lower() == "yes"
cough = input("Do you have cough? ").lower() == "yes"
headache = input("Do you have headache? ").lower() == "yes"
sore_throat = input("Do you have sore throat? ").lower() == "yes"

# Diagnosis
result = diagnose(fever, cough, headache, sore_throat)

# Display result
print("\n===== Diagnosis Result =====")
print("Possible Disease:", result)
print("\nNote: Please consult a qualified medical professional for proper diagnosis.")
```
